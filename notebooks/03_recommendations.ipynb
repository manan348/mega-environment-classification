# Phase 5: Recommendation Layer

Create templates for variety/practice recommendations per cluster, then build the final lookup.

```python
import sys
sys.path.insert(0, '../src')

import pandas as pd
from mega_env.config import load_config
from mega_env.recommendations import (
    create_recommendation_template,
    build_recommendation_lookup,
    save_recommendation_lookup
)

config = load_config('../config/settings.yaml')
```

```python
# Load district clusters
district_clusters = pd.read_csv('../data/processed/district_clusters.csv')
district_col = [c for c in district_clusters.columns if 'NAME' in c.upper() or 'district' in c.lower()][0]
print(f"District column: {district_col}")
print(f"Clusters: {sorted(district_clusters['cluster'].unique())}")
district_clusters[[district_col, 'cluster']].head(10)
```

```python
# Create templates
variety_df, practice_df = create_recommendation_template(
    district_clusters, district_col, config.output_paths['recommendations']
)
```

## Now edit the CSV files manually:

1. **`data/processed/recommendations_varieties.csv`** - Add recommended wheat varieties per cluster
2. **`data/processed/recommendations_practices.csv`** - Add agronomic practices per cluster

Sources for this data:
- Punjab Agriculture Department zone recommendations
- Wheat variety trial results (PARC, AARI, etc.)
- Extension service records
- Published literature on Punjab wheat zones

After editing, run the next cell to build the final lookup.

```python
# Build final lookup (run after editing CSVs)
lookup = build_recommendation_lookup(
    '../data/processed/recommendations_varieties.csv',
    '../data/processed/recommendations_practices.csv'
)
```

```python
# Save lookup
save_recommendation_lookup(lookup, config.output_paths['recommendations'])
```

```python
# Preview lookup
import json
with open(config.output_paths['recommendations']) as f:
    data = json.load(f)

for cluster, recs in data.items():
    print(f"\nCluster {cluster}:")
    print(f"  Varieties: {recs['varieties']}")
    print(f"  Practices: {recs['practices']}")
```
