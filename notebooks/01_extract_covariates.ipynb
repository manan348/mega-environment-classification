# Phase 1: Covariate Extraction

Extract weather (ERA5), soil (SoilGrids), and topography (SRTM) covariates for all Punjab districts.

```python
import sys
sys.path.insert(0, '../src')

from mega_env.config import load_config
from mega_env.gee_utils import run_full_extraction, load_district_boundaries, initialize_gee

config = load_config('../config/settings.yaml')
```

```python
# Initialize GEE
initialize_gee(config.gee_project_id)
```

```python
# Load districts
districts = load_district_boundaries(config.districts_boundary_file)
districts.head()
```

```python
# Run extraction (this may take several minutes)
covariate_df = run_full_extraction(config, districts)
```

```python
# Inspect results
print(f"Shape: {covariate_df.shape}")
print(f"Columns: {list(covariate_df.columns)}")
print(f"Years: {sorted(covariate_df['year'].unique())}")
covariate_df.head()
```

```python
# Check for missing values
missing = covariate_df.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

```python
# Save
covariate_df.to_parquet('../data/processed/covariate_table.parquet', index=False)
print("✓ Saved to data/processed/covariate_table.parquet")
```
